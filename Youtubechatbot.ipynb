```python
from youtube_transcript_api import YouTubeTranscriptApi
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_community.vectorstores import FAISS
```

```python
video_id = "J5_-l7WIO_w"
api = YouTubeTranscriptApi()

transcript_list = api.fetch(
    video_id,
    languages=["hi"]
 )

transcript = " ".join(
    snippet.text for snippet in transcript_list
 )
print(transcript)
```

```python
splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000,
    chunk_overlap=200
)
chunks = splitter.create_documents([transcript])

print("Number of chunks:", len(chunks))
```

```python
 embeddings = HuggingFaceEmbeddings(
      model_name="sentence-transformers/paraphrase-multilingual-MiniLM-L12-v2"
       )
```

```python
vector_store = FAISS.from_documents(
    chunks,
    embeddings
 )
```

```python
retriever = vector_store.as_retriever(
     search_type="similarity",
     search_kwargs={"k": 4}
)
```

```python
results = retriever.invoke(
  "What is DeepMind?"
      )
```

```python
for doc in results:
     print(doc.page_content)
```

```python
question = "is the topic of aliens discussed in this video? if yes then what was discussed"

retrieved_docs = retriever.invoke(question)
```

```python
context_text = "\n\n".join(
    doc.page_content for doc in retrieved_docs
)
```

```python
from langchain_core.prompts import PromptTemplate
```

```python
from langchain_huggingface import HuggingFacePipeline, ChatHuggingFace
from transformers import pipeline

LLM_MODEL = "Qwen/Qwen2.5-1.5B-Instruct"
pipe = pipeline(
    "text-generation",
    model=LLM_MODEL,
    max_new_tokens=256,
    return_full_text=False
)
base_llm = HuggingFacePipeline(pipeline=pipe)
llm = ChatHuggingFace(llm=base_llm, model_id=LLM_MODEL, tokenizer=pipe.tokenizer)
```

```python
prompt = PromptTemplate(
    template="""
    You are a helpful assistant.
    Answer ONLY from the provided transcript context.
    If the context is insufficient, just say you don't know.

    {context}
    Question: {question}
    """,
)
```

```python
final_prompt = prompt.invoke({
    "context": context_text,
    "question": question
})
```

```python
answer = llm.invoke(final_prompt)

print(answer)
```

Chain

```python
from langchain_core.runnables import RunnableParallel, RunnablePassthrough, RunnableLambda
from langchain_core.output_parsers import StrOutputParser
```

```python
def format_docs(retrieved_docs):
   context_text = "\n\n".join(doc.page_content for doc in retrieved_docs)
   return context_text
```

```python
parallel_chain = RunnableParallel({
    'context': retriever | RunnableLambda(format_docs),
    'question': RunnablePassthrough()
})
```

```python
parallel_chain.invoke("who is Demis")
```

```python
parser = StrOutputParser()
```

```python
main_chain = parallel_chain | prompt | llm | parser
```

```python
main_chain.invoke("Can you summarize the video")
```

```python
import os, joblib

SAVE_DIR = "saved_files"
os.makedirs(SAVE_DIR, exist_ok=True)

# 1) FAISS index -> creates index.faiss and index.pkl
vector_store.save_local(f"{SAVE_DIR}/faiss_index")

# 2) Transcript chunks
joblib.dump(chunks, f"{SAVE_DIR}/chunks.pkl")

# 3) Settings used to build the index (read from your own variables)
metadata = {
    "video_id": video_id,
     "embedding_model": embeddings.model_name,
     "llm_model": LLM_MODEL,
     "chunk_size": 1000,
     "chunk_overlap": 200,
     "k": 4,
     "num_chunks": len(chunks),
     }
joblib.dump(metadata, f"{SAVE_DIR}/metadata.pkl")

print("Saved:", os.listdir(SAVE_DIR))
print(metadata)
```

```python
from langchain_community.vectorstores import FAISS

loaded_store = FAISS.load_local(
    f"{SAVE_DIR}/faiss_index", embeddings, allow_dangerous_deserialization=True
)
loaded_chunks = joblib.load(f"{SAVE_DIR}/chunks.pkl")
print("Chunks reloaded:", len(loaded_chunks))
print(loaded_store.similarity_search("aliens", k=1)[0].page_content[:200])
```

```python
import hashlib
from importlib.metadata import version

pkgs = ["youtube-transcript-api", "langchain", "langchain-core", "langchain-community",
        "langchain-huggingface", "langchain-text-splitters", "faiss-cpu",
         "sentence-transformers", "transformers", "joblib"]
lines = []
for p in pkgs:
  try:
    lines.append(f"{p}=={version(p)}")
  except Exception:
    lines.append(p)
lines.append("torch")
open("requirements.txt", "w").write("\n".join(lines) + "\n")
print(open("requirements.txt").read())

print("Fingerprints (SHA-256, first 16 characters):")
for root, _, fnames in os.walk(SAVE_DIR):
  for f in sorted(fnames):
    path = os.path.join(root, f)
    print(hashlib.sha256(open(path, "rb").read()).hexdigest()[:16], path)
```
